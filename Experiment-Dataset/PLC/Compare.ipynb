```python
from tqdm import tqdm

def levenshtein_distance(s1, s2):
    if len(s1) < len(s2):
        return levenshtein_distance(s2, s1)

    if len(s2) == 0:
        return len(s1)

    previous_row = range(len(s2) + 1)
    for i, c1 in tqdm(enumerate(s1), total=len(s1)):
        current_row = [i + 1]
        for j, c2 in enumerate(s2):
            insertions = previous_row[j + 1] + 1
            deletions = current_row[j] + 1
            substitutions = previous_row[j] + (c1 != c2)
            current_row.append(min(insertions, deletions, substitutions))
        previous_row = current_row

    return previous_row[-1]
    
print(levenshtein_distance(sim_pcs,real_pcs))
```

```python
from tqdm import trange

def lcs(X, Y):
    m = len(X)
    n = len(Y)
    L = [[0] * (n + 1) for _ in range(m + 1)]

    for i in trange(1, m + 1):
        for j in range(1, n + 1):
            if X[i-1] == Y[j-1]:
                L[i][j] = L[i-1][j-1] + 1
            else:
                L[i][j] = max(L[i-1][j], L[i][j-1])

    return L[m][n]

def lcs_distance(X, Y):
    lcs_length = lcs(X, Y)
    print(f'lcs_length {lcs_length}')
    return len(X) + len(Y) - 2 * lcs_length

# 示例使用
seq1 = 'ABCD'
seq2 = 'ABDE'

# lcs_len = lcs(seq1, seq2)
lcs_dist = lcs_distance(seq1, seq2)

print(f"序列1: {seq1}")
print(f"序列2: {seq2}")
print(f"LCS distance: {lcs_dist}")
```
